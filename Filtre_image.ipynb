Code qui suit sert à appliquer le critère de netteté à toutes les images et les énregistré dans le dossier image_net

```python

import numpy as np
import os
import cv2
```

```python
# 1. Définir les dossiers source et de destination
dossier_source = "./bee_dataset_tiled/images"  # Dossier contenant vos images (ex: "M01C02_...")
dossier_destination = "./bee_dataset_tiled/images_net" # Dossier où seront enregistrées les images modifiées

# Créer le dossier de destination s'il n'existe pas
os.makedirs(dossier_destination, exist_ok=True)

# Définir le noyau (kernel) d'accentuation initial
kernel_sharpening = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
])

images_traitees = 0

# Parcourir tous les fichiers du dossier source
for nom_fichier in os.listdir(dossier_source):
    # Filtrer pour s'assurer de ne traiter que les images
    if nom_fichier.lower().endswith(('.jpg', '.jpeg', '.png')):
        chemin_image_source = os.path.join(dossier_source, nom_fichier)
        
        # Charger l'image
        image = cv2.imread(chemin_image_source)
        
        if image is not None:
            # Appliquer le filtre de netteté à l'aide de la matrice
            image_nette = cv2.filter2D(image, -1, kernel_sharpening)
            
            # Séparer le nom du fichier et son extension
            nom_base, extension = os.path.splitext(nom_fichier)
            
            # Créer le nouveau nom avec le suffixe "_nettete"
            nouveau_nom = f"{nom_base}_nettete{extension}"
            chemin_image_destination = os.path.join(dossier_destination, nouveau_nom)
            
            # Sauvegarder l'image traitée
            cv2.imwrite(chemin_image_destination, image_nette)
            images_traitees += 1

print(f"Terminé ! {images_traitees} images ont été traitées avec la matrice de netteté et enregistrées dans le dossier '{dossier_destination}'.")
```

Doc qui suit permet d'avoir les images filtrées CLAHE

```python
dossier_source = "./bee_dataset_tiled/images"  # Dossier contenant vos images
dossier_destination = "./bee_dataset_tiled/images_clahe"  # Dossier où seront enregistrées les images modifiées

# Créer le dossier de destination s'il n'existe pas
os.makedirs(dossier_destination, exist_ok=True)

# 2. Configurer le filtre CLAHE
# clipLimit : limite le contraste pour éviter le bruit excessif (ex: 2.0 ou 3.0)
# tileGridSize : taille de la grille pour l'égalisation locale (ex: 8x8)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

images_traitees = 0

for nom_fichier in os.listdir(dossier_source):
    if nom_fichier.lower().endswith(('.jpg', '.jpeg', '.png')):
        chemin_image_source = os.path.join(dossier_source, nom_fichier)
        
        # Charger l'image directement en nuances de gris (grâce au paramètre 0)
        image = cv2.imread(chemin_image_source, cv2.IMREAD_GRAYSCALE)
        
        if image is not None:
            # Appliquer CLAHE directement sur l'image en nuances de gris
            image_clahe = clahe.apply(image)
            
            # Séparer le nom du fichier et son extension
            nom_base, extension = os.path.splitext(nom_fichier)
            
            # Créer le nouveau nom avec le suffixe "_clahe"
            nouveau_nom = f"{nom_base}_clahe{extension}"
            chemin_image_destination = os.path.join(dossier_destination, nouveau_nom)
            
            # Sauvegarder l'image traitée
            cv2.imwrite(chemin_image_destination, image_clahe)
            images_traitees += 1

print(f"Terminé ! {images_traitees} images ont été traitées avec le filtre CLAHE.")
```

```python
dossier_source = "./bee_dataset_tiled/images"  
dossier_destination = "./bee_dataset_tiled/images_median"  

os.makedirs(dossier_destination, exist_ok=True)

# Définir la taille du noyau (doit être un nombre impair, par exemple 3, 5, etc.)
# Plus la valeur est grande, plus le lissage est fort.
taille_noyau = 3

images_traitees = 0

# Parcourir tous les fichiers du dossier source
for nom_fichier in os.listdir(dossier_source):
    if nom_fichier.lower().endswith(('.jpg', '.jpeg', '.png')):
        chemin_image_source = os.path.join(dossier_source, nom_fichier)
        
        # Charger l'image directement en nuances de gris
        image = cv2.imread(chemin_image_source, cv2.IMREAD_GRAYSCALE)
        
        if image is not None:
            # Appliquer le filtre médian avec cv2.medianBlur
            image_median = cv2.medianBlur(image, taille_noyau)
            
            # Séparer le nom du fichier et son extension
            nom_base, extension = os.path.splitext(nom_fichier)
            
            # Créer le nouveau nom avec le suffixe "_median"
            nouveau_nom = f"{nom_base}_median{extension}"
            chemin_image_destination = os.path.join(dossier_destination, nouveau_nom)
            
            # Sauvegarder l'image traitée
            cv2.imwrite(chemin_image_destination, image_median)
            images_traitees += 1

print(f"Terminé ! {images_traitees} images ont été traitées avec le filtre médian.")
```
